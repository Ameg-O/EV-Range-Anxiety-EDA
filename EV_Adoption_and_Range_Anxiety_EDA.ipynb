## 28. IQR limits for annual income



```python
Q1=df['Annual_Income_USD'].quantile(0.25)
Q3=df['Annual_Income_USD'].quantile(0.75)

IQR=Q3-Q1

lower=Q1-1.5*IQR
higher=Q3+1.5*IQR

print("Lower limit",lower)
print("Higher limit",higher)

income_outiler=df[(df['Annual_Income_USD']<lower)|
                  (df['Annual_Income_USD']>higher)]
print("number of outliers",len(income_outiler))
```

```python
sns.boxplot(data=df, x="Daily_Commute_km")
plt.title("Daily_commute Outlier Analysis")
plt.show()
```

## 29. IQR limits for commute distance


```python
Q1=df['Daily_Commute_km'].quantile(0.25)
Q3=df['Daily_Commute_km'].quantile(0.75)

IQR=Q3-Q1

lower=Q1-1.5*IQR
higher=Q3+1.5*IQR

print("Lower limit",lower)
print("Higher limit",higher)

commute_outiler=df[(df['Daily_Commute_km']<lower)|
                  (df['Daily_Commute_km']>higher)]
print("number of outliers",len(commute_outiler))
```

## 37. Correlation heatmap

Select numerical columns, exclude the buyer ID, calculate correlations, and display them as a heatmap. Values near +1 indicate positive linear association, values near −1 negative association, and values near 0 little linear association. Correlation does not prove causation.

```python
df1 = df.select_dtypes(include='number')

df1 = df1.drop(
    columns=['Buyer_ID'],
    errors='ignore'
)

correlation = df1.corr()

plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Heatmap')
plt.show()
```

Key Insights
- Among people with high range anxiety, only 0.60% were willing to buy an EV.
- Among people with low range anxiety, 19.87% were willing to buy an EV.
- 27.45% of people with subsidies available were willing to buy an EV, compared with 2.53% without subsidies.
- Range anxiety showed a strong association with EV purchase intention.
- Subsidy availability was associated with higher willingness to purchase an EV.

- The IQR method identified 58 income outliers and 44 daily commute-distance outliers.
